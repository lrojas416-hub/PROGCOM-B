```python
def limpiar(palabra):
    palabra = palabra.strip().lower()
    for con, sin in zip("áéíóú", "aeiou"):
        palabra = palabra.replace(con, sin)
    return palabra

D
def es_valida(palabra, letra):
    palabra = limpiar(palabra)
    if len(palabra) < 2:
        return False
    if not palabra.replace(" ", "").isalpha():
        return False
    return palabra[0] == limpiar(letra)
```

```python
def respuestas_maquina(letra, dificultad):
    respuestas = {}
    for cat in categorias:
        palabras = banco[letra][cat]
        if random.random() < dificultades[dificultad]["prob"]:
            if dificultad == "Fácil":
                respuesta = random.choice(palabras[:2])
            elif dificultad == "Normal":
                respuesta = random.choice(palabras)
            else:
                respuesta = random.choice(palabras[2:])
        else:
            respuesta = ""
        respuestas[cat] = respuesta
    return respuestas
```

```python
def calcular_puntos(letra, resp_yo, resp_pc):
    pts_yo = {}
    pts_pc = {}
    for cat in categorias:
        yo = resp_yo[cat]
        pc = resp_pc[cat]
        ok_yo = es_valida(yo, letra)
        ok_pc = es_valida(pc, letra)
        repetida = ok_yo and ok_pc and limpiar(yo) == limpiar(pc)

        pts_yo[cat] = 0
        pts_pc[cat] = 0
        if ok_yo:
            pts_yo[cat] = 5 if repetida else 10
        if ok_pc:
            pts_pc[cat] = 5 if repetida else 10
    return pts_yo, pts_pc


def sumar(historial):
    total = 0
    for ronda in historial:
        total = total + ronda["total"]
    return total
```

```python
estilo = "border:2px solid #c0392b; padding:6px; text-align:center;"


def hacer_tabla(nombre, rondas):
    t = "<b>" + nombre + "</b>"
    t += "<table style='border-collapse:collapse; background:#fdfbf3; color:#222; font-family:Comic Sans MS, cursive;'>"
    t += "<tr><th style='" + estilo + "'>Letra</th>"
    for cat in categorias:
        t += "<th style='" + estilo + "'>" + cat + "</th>"
    t += "<th style='" + estilo + "'>Puntos</th></tr>"

    for r in rondas:
        t += "<tr><td style='" + estilo + "'>" + r["letra"] + "</td>"
        for cat in categorias:
            palabra = html.escape(r["resp"][cat])
            t += "<td style='" + estilo + "'>" + palabra + "<br>" + str(r["pts"][cat]) + "</td>"
        t += "<td style='" + estilo + "'><b>" + str(r["total"]) + "</b></td></tr>"

    t += "</table><br>"
    return t
```

```python
pantalla = widgets.VBox()
juego = {}


def menu(boton=None):
    titulo = widgets.HTML("<h2>STOP</h2><p>Juega contra la máquina. Elige la dificultad.</p>")
    lista = widgets.Dropdown(options=list(dificultades), value="Normal", description="Dificultad:")
    boton_empezar = widgets.Button(description="Empezar", button_style="success")
    boton_empezar.on_click(empezar)
    juego["lista"] = lista
    pantalla.children = [titulo, lista, boton_empezar]


def empezar(boton):
    juego["dificultad"] = juego["lista"].value
    juego["ronda"] = 0
    juego["letras_usadas"] = []
    juego["historial_yo"] = []
    juego["historial_pc"] = []
    nueva_ronda(None)


def nueva_ronda(boton):
    juego["ronda"] = juego["ronda"] + 1

    letra = random.choice(list(banco))
    while letra in juego["letras_usadas"]:
        letra = random.choice(list(banco))
    juego["letras_usadas"].append(letra)
    juego["letra"] = letra

    juego["resp_pc"] = respuestas_maquina(letra, juego["dificultad"])
    juego["inicio"] = time.time()

    juego["cajas"] = {}
    cajas = []
    for cat in categorias:
        caja = widgets.Text(description=cat)
        juego["cajas"][cat] = caja
        cajas.append(caja)

    tiempo = dificultades[juego["dificultad"]]["tiempo"]
    titulo = widgets.HTML("<h2>Ronda " + str(juego["ronda"]) + " de " + str(RONDAS)
                          + " - Letra: " + letra + "</h2>")
    info = widgets.HTML("Dificultad: " + juego["dificultad"] + ". Tienes " + str(tiempo)
                        + " segundos antes de que la máquina grite STOP.")
    boton_stop = widgets.Button(description="STOP!", button_style="danger")
    boton_tiempo = widgets.Button(description="Ver mi tiempo")
    boton_stop.on_click(gritar_stop)
    boton_tiempo.on_click(ver_tiempo)
    juego["aviso"] = widgets.HTML("")

    pantalla.children = [titulo, info] + cajas + [widgets.HBox([boton_stop, boton_tiempo]), juego["aviso"]]


def ver_tiempo(boton):
    segundos = int(time.time() - juego["inicio"])
    juego["aviso"].value = "Llevas " + str(segundos) + " segundos"


def gritar_stop(boton):
    resp_yo = {}
    for cat in categorias:
        resp_yo[cat] = juego["cajas"][cat].value
        if resp_yo[cat].strip() == "":
            juego["aviso"].value = "Te falta llenar: " + cat
            return

    letra = juego["letra"]
    pts_yo, pts_pc = calcular_puntos(letra, resp_yo, juego["resp_pc"])

    segundos = int(time.time() - juego["inicio"])
    limite = dificultades[juego["dificultad"]]["tiempo"]
    if segundos > limite:
        mensaje = "Te demoraste " + str(segundos) + " s y la máquina gritó STOP primero. No sumas puntos."
        for cat in categorias:
            pts_yo[cat] = 0
    else:
        mensaje = "Gritaste STOP a los " + str(segundos) + " s."

    juego["historial_yo"].append({"letra": letra, "resp": resp_yo, "pts": pts_yo, "total": sum(pts_yo.values())})
    juego["historial_pc"].append({"letra": letra, "resp": juego["resp_pc"], "pts": pts_pc, "total": sum(pts_pc.values())})
    mostrar_resultado(mensaje)


def mostrar_resultado(mensaje):
    tabla_yo = hacer_tabla("Tú", [juego["historial_yo"][-1]])
    tabla_pc = hacer_tabla("Máquina", [juego["historial_pc"][-1]])
    total = "Total: Tú " + str(sumar(juego["historial_yo"])) + " - Máquina " + str(sumar(juego["historial_pc"]))

    if juego["ronda"] == RONDAS:
        boton = widgets.Button(description="Ver resultado final", button_style="primary")
        boton.on_click(mostrar_final)
    else:
        boton = widgets.Button(description="Siguiente ronda", button_style="primary")
        boton.on_click(nueva_ronda)

    texto = "<h3>Ronda " + str(juego["ronda"]) + "</h3><p>" + mensaje + "</p>" + tabla_yo + tabla_pc + "<p>" + total + "</p>"
    pantalla.children = [widgets.HTML(texto), boton]


def mostrar_final(boton):
    total_yo = sumar(juego["historial_yo"])
    total_pc = sumar(juego["historial_pc"])
    if total_yo > total_pc:
        resultado = "Ganaste!"
    elif total_yo < total_pc:
        resultado = "Ganó la máquina, revancha?"
    else:
        resultado = "Empate"

    texto = "<h2>Fin del juego: " + resultado + "</h2>"
    texto += "<p>Tú: " + str(total_yo) + " puntos. Máquina: " + str(total_pc) + " puntos.</p>"
    texto += hacer_tabla("Tú", juego["historial_yo"])
    texto += hacer_tabla("Máquina", juego["historial_pc"])

    boton_otra = widgets.Button(description="Jugar otra vez", button_style="success")
    boton_otra.on_click(menu)
    pantalla.children = [widgets.HTML(texto), boton_otra]
```
